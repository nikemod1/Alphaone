```python
# ================= INSTALL =================
!pip install -q segmentation-models-pytorch albumentations opencv-python

# ================= IMPORTS =================
import os, cv2, torch, numpy as np
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import albumentations as A
import segmentation_models_pytorch as smp
from tqdm import tqdm

# ================= CONFIG =================
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

IMG_SIZE = 384
BATCH_SIZE = 8
EPOCHS = 10
NUM_CLASSES = 11

torch.backends.cudnn.benchmark = True

# ================= PATH =================
BASE = "/content/Offroad_Segmentation_Training_Dataset"

train_img = f"{BASE}/train/Color_Images"
train_mask = f"{BASE}/train/Segmentation"

val_img = f"{BASE}/val/Color_Images"
val_mask = f"{BASE}/val/Segmentation"

print("Train:", len(os.listdir(train_img)))
print("Val:", len(os.listdir(val_img)))

# ================= LABEL MAP =================
LABEL_MAP = {
    0: 0,
    100: 1, 200: 2, 300: 3, 500: 4,
    550: 5, 600: 6, 700: 7, 800: 8,
    7100: 9, 10000: 10
}

def remap(mask):
    new = np.zeros_like(mask)
    for k, v in LABEL_MAP.items():
        new[mask == k] = v
    return new

# ================= DATASET =================
class SegDataset(Dataset):
    def __init__(self, img_dir, mask_dir, train=True):
        self.imgs = sorted(os.listdir(img_dir))
        self.img_dir = img_dir
        self.mask_dir = mask_dir

        if train:
            self.tf = A.Compose([
                A.Resize(IMG_SIZE, IMG_SIZE),
                A.HorizontalFlip(p=0.5),
                A.RandomBrightnessContrast(p=0.3),
                A.GaussNoise(p=0.2),
            ])
        else:
            self.tf = A.Compose([
                A.Resize(IMG_SIZE, IMG_SIZE),
            ])

    def __len__(self):
        return len(self.imgs)

    def __getitem__(self, i):
        img = cv2.imread(os.path.join(self.img_dir, self.imgs[i]))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        mask = cv2.imread(os.path.join(self.mask_dir, self.imgs[i]), 0)
        mask = remap(mask)

        aug = self.tf(image=img, mask=mask)
        img, mask = aug["image"], aug["mask"]

        img = torch.tensor(img).permute(2,0,1).float()/255
        mask = torch.tensor(mask).long()

        return img, mask

# ================= LOADERS =================
train_loader = DataLoader(SegDataset(train_img, train_mask, True),
                          batch_size=BATCH_SIZE, shuffle=True)

val_loader = DataLoader(SegDataset(val_img, val_mask, False),
                        batch_size=BATCH_SIZE)

# ================= MODEL =================
model = smp.Unet("resnet34", encoder_weights="imagenet",
                 classes=NUM_CLASSES).to(DEVICE)

# ================= LOSS =================
weights = torch.tensor([0.05] + [1]*10).to(DEVICE)
ce_loss = nn.CrossEntropyLoss(weight=weights)

def dice_loss(pred, target):
    pred = torch.softmax(pred, dim=1)
    target = F.one_hot(target, NUM_CLASSES).permute(0,3,1,2).float()

    inter = (pred * target).sum((2,3))
    union = pred.sum((2,3)) + target.sum((2,3))

    return 1 - ((2*inter + 1e-6) / (union + 1e-6)).mean()

def loss_fn(pred, target):
    return ce_loss(pred, target) + dice_loss(pred, target)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)

# 🔥 LR scheduler (reduces overfitting)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode='max', patience=2, factor=0.5
)

# ================= METRIC =================
def iou_score(pred, mask):
    pred = torch.argmax(pred, dim=1)
    ious = []

    for cls in range(NUM_CLASSES):
        p = (pred == cls)
        t = (mask == cls)

        inter = (p & t).sum().float()
        union = (p | t).sum().float()

        if union > 0:
            ious.append((inter+1e-6)/(union+1e-6))

    return torch.mean(torch.stack(ious))

# ================= TRAIN =================
best_iou = 0
patience = 3
no_improve = 0

for epoch in range(EPOCHS):

    model.train()
    total_loss = 0

    for x,y in tqdm(train_loader):
        x,y = x.to(DEVICE), y.to(DEVICE)

        pred = model(x)
        loss = loss_fn(pred, y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    # ===== VALIDATION =====
    model.eval()
    val_iou = 0

    with torch.no_grad():
        for x,y in val_loader:
            x,y = x.to(DEVICE), y.to(DEVICE)
            pred = model(x)

            val_iou += iou_score(pred,y).item()

    val_iou /= len(val_loader)

    print(f"\nEpoch {epoch+1}")
    print(f"Loss: {total_loss/len(train_loader):.4f}")
    print(f"IoU: {val_iou:.4f}")

    scheduler.step(val_iou)

    if val_iou > best_iou:
        best_iou = val_iou
        torch.save(model.state_dict(), "best_model.pth")
        print("✅ Saved Best Model")
        no_improve = 0
    else:
        no_improve += 1

    # 🔥 Early stopping
    if no_improve >= patience:
        print("⛔ Early stopping triggered")
        break

print("\n🎉 Training Done")
```
